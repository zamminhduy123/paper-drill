# 03 — Gap/Idea Probe
Feed all limitations → Qwen → blind spots + cross-pollination. Run on server.

```python
import os, requests

BASE_URL=os.environ.get("LLAMA_BASE_URL", " https://divorcee-work-pessimism.ngrok-free.dev/v1")
MODEL=os.environ.get("LLAMA_MODEL","Swift-Qwen3.8-27B-Q4_K_M")
THESIS="We study the application of deep learning methods into areas of In-Vehicle Network"
LIMS="""1.KD: open-set degrades in student; no cross-dataset\n2.SLM: tokenization loses numerics; rare attacks weak\n3.OpenSet: needs labeled target samples; no drift model\n4.Explainable: rules fail rare/unseen; NMF linear\n5.Synthetic: may break protocol semantics; no HW val\n6.LLM/VLM: not ECU-deployable; cloud needed\n7.ViT: loses temporal precision; heavy\n8.Multimodal: needs synced data; expensive\n9.GNN+LLM zero-shot: descriptions miss protocol semantics\n10.EdgeAI: compression hurts open-set/rare"""
PROMPT=f"Thesis: {THESIS}\nLimitations:\n{{}}\n\nOutput:\n## Blind Spots (2)\n## Cross-Pollination (2: Method X from A -> Problem Y from B)\n## Testable Hypotheses (2-3, with dataset + metric)"
print("ready")
```

```python
r = requests.post(f"{BASE_URL}/chat/completions",json={"model":MODEL,"messages":[{"role":"user","content":PROMPT.format(LIMS)}],"temperature":0.7},timeout=300)
```
